# Dataset Exploration

Quick checks for the available raw ergonomic posture datasets.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path('..').resolve()
RAW = ROOT / 'data' / 'raw'
print(RAW)
for path in RAW.iterdir():
    print(path.name, 'dir' if path.is_dir() else 'file')
```

```python
multiposture = pd.read_csv(RAW / 'multiposture' / 'data.csv')
print(multiposture.shape)
display(multiposture.head())
display(multiposture[['upperbody_label', 'lowerbody_label']].value_counts().head(20))
```

```python
office_root = RAW / 'office_posture' / 'archives_data'
office_labels = pd.read_csv(office_root / 'labels' / 'labels_for_train.csv')
print(office_labels.shape)
display(office_labels['label'].value_counts())
sample_json = next((office_root / 'keypoints').glob('*.json'))
payload = json.loads(sample_json.read_text())
print(sample_json.name, len(payload[0]['keypoints']), 'keypoints')
```
