# Model Training And Evaluation

Train SVM and Random Forest, save the best model, then generate the model evaluation image with accuracy, confusion matrix, per-class metrics, and feature importance.

```python
from pathlib import Path
import sys

import pandas as pd
from IPython.display import Image, display

ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT))

from scripts.train_svm import train
from scripts.evaluate_model import evaluate
```

## Train And Compare SVM vs Random Forest

```python
dataset_path = ROOT / 'data' / 'processed' / 'dataset_final.csv'
svm_model_path = ROOT / 'models' / 'svm_model.pkl'
best_model_path = ROOT / 'models' / 'best_model.pkl'
results_dir = ROOT / 'results'

train_metrics = train(dataset_path, svm_model_path, results_dir, best_model_path)
print(f"Best model: {train_metrics['model_name']}")
print(f"Best accuracy: {train_metrics['accuracy']:.4f}")
display(pd.DataFrame(train_metrics['model_comparison']).T)
```

## Evaluate Best Model

```python
evaluation_png = ROOT / 'results' / 'model_evaluation.png'
metrics = evaluate(dataset_path, best_model_path, evaluation_png)

print(f"Overall accuracy: {metrics['accuracy']:.4f}")
print('Confusion matrix:')
display(pd.DataFrame(metrics['confusion_matrix'], index=metrics['labels'], columns=metrics['labels']))
print('Per-class precision / recall / F1:')
display(pd.DataFrame(metrics['classification_report']).T.loc[metrics['labels'], ['precision', 'recall', 'f1-score']])
print('Feature importance:')
display(pd.DataFrame(metrics['feature_importance']))
display(Image(filename=str(evaluation_png)))
```
